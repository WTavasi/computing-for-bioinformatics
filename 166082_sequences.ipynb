## Part 1: Demonstration — TP53 (Human)

TP53 is a tumor-suppressor gene thathalts cell division when DNA is damaged.

**Cell 1:** install the Biopython library (needed once per Colab session).

```python
!pip install biopython
```

**Cell 2:** import the tools we need, and set your email — NCBI requires this on every request.

```python
from Bio import Entrez, SeqIO
Entrez.email = "wyclef.tavasi@strathmore.edu"
```

**Cell 3:** search the NCBI *nucleotide* database for TP53 in human, and get back a list of matching record IDs.

```python
handle = Entrez.esearch(db="nucleotide", term="TP53[Gene] AND Homo sapiens[Organism]", retmax=3)
record = Entrez.read(handle)
handle.close()

print("Found IDs:", record["IdList"])
```

**Cell 4:** fetch the actual sequence for one of those IDs, in FASTA format.

> **Note:** If you get your output as series of N this is not an error, it shows that there is a gap in the sequence. Select another record from the IdList

```python
id_to_fetch = record["IdList"][2]
handle = Entrez.efetch(db="nucleotide", id=id_to_fetch, rettype="fasta", retmode="text")
seq_record = SeqIO.read(handle, "fasta")
handle.close()

print("Sequence ID:", seq_record.id)
print("Sequence Length:", len(seq_record.seq))
print("First 100 bases:", seq_record.seq[:100])
```

**Cell 5:** compute GC content — the percentage of the sequence that is G or C.

**Why it matters:** GC content gives a rough first impression of how two sequences compare.

```python
gc_content = 100 * float(seq_record.seq.count("G") + seq_record.seq.count("C")) / len(seq_record.seq)
print("GC Content: %.2f%%" % gc_content)
```

**Cell 6:** save the sequence to a FASTA file.

```python
SeqIO.write([seq_record], "TP53_sequence.fasta", "fasta")
print("Saved to TP53_sequence.fasta")
```

## Part 2: LAB ASSIGNMENT — COX-2 (PTGS2) in Human vs. Fish

COX-2 is the enzyme targeted by pain-relief drugs like ibuprofen.
Its official NCBI gene symbol is **PTGS2** .You will not get reliable results if you search "COX2".

Repeat Cells 3–5 from Part 1 **twice** — once for human, once for zebrafish (*Danio rerio*) — then combine both sequences into **one** FASTA file.

Use these search terms:

| Species | Search term (for Cell 3) |
|---|---|
| Human | `PTGS2[Gene] AND Homo sapiens[Organism]` |
| Zebrafish | `PTGS2[Gene] AND Danio rerio[Organism]` |


### Human COX-2 — search, fetch, GC content


```python
human_handle = Entrez.esearch(db="nucleotide",
    term="PTGS2[Gene] AND Homo sapiens[Organism] AND RefSeq[filter] AND biomol_mrna[PROP]",
    retmax=3)
human_search = Entrez.read(human_handle)
human_handle.close()

print("Human IDs found:", human_search["IdList"])
```

```python
# I used the reference mRNA accession so that I get the same sequence each time
human_handle = Entrez.efetch(db="nucleotide", id="NM_000963.4",
                              rettype="fasta", retmode="text")
human_cox2 = SeqIO.read(human_handle, "fasta")
human_handle.close()

print("Human sequence ID:", human_cox2.id)
print("Human sequence length:", len(human_cox2.seq))
print("First 100 bases:", human_cox2.seq[:100])
```

```python
human_gc = 100 * float(human_cox2.seq.count("G") + human_cox2.seq.count("C")) / len(human_cox2.seq)
print("Human GC Content: %.2f%%" % human_gc)
```

### Fish COX-2 — search, fetch, GC content

Zebrafish have a COX-2 gene named ptgs2a. The search term from the manual still finds this record.


```python
fish_handle = Entrez.esearch(db="nucleotide",
    term="PTGS2[Gene] AND Danio rerio[Organism] AND RefSeq[filter] AND biomol_mrna[PROP]",
    retmax=3)
fish_search = Entrez.read(fish_handle)
fish_handle.close()

print("Fish IDs found:", fish_search["IdList"])
```

```python
fish_handle = Entrez.efetch(db="nucleotide", id="NM_153657.1",
                             rettype="fasta", retmode="text")
fish_cox2 = SeqIO.read(fish_handle, "fasta")
fish_handle.close()

print("Fish sequence ID:", fish_cox2.id)
print("Fish sequence length:", len(fish_cox2.seq))
print("First 100 bases:", fish_cox2.seq[:100])
```

```python
fish_gc = 100 * float(fish_cox2.seq.count("G") + fish_cox2.seq.count("C")) / len(fish_cox2.seq)
print("Fish GC Content: %.2f%%" % fish_gc)
```

### Combine both sequences into a FASTA file (Multi-FASTA file)

`SeqIO.write()` accepts a list of records, so both sequences can be written into the same file.


```python
cox2_records = [human_cox2, fish_cox2]
number_saved = SeqIO.write(cox2_records, "COX2_human_fish.fasta", "fasta")

print("Number of sequences saved:", number_saved)
print("Saved to COX2_human_fish.fasta")
```

```python
saved_records = list(SeqIO.parse("COX2_human_fish.fasta", "fasta"))
for item in saved_records:
    print(item.id, "-", len(item.seq), "bases")
```
